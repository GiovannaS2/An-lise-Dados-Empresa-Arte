```python
import pandas as pd
import matplotlib.pyplot as plt

# --- REPETINDO A BASE DE DADOS ---
data = {
    'ID_Pedido': [101, 102, 103, 104, 105],
    'Produto': ['Reborn Luxo', 'Escultura BJD', 'Action Figure', 'Reborn Luxo', 'Acessório Silicone'],
    'Preco_Venda': [2500, 3200, 1500, 2500, 450],
    'Custo_Producao': [800, 1100, 500, 800, 120]
}
df = pd.DataFrame(data)
df['Lucro_Liquido'] = df['Preco_Venda'] - df['Custo_Producao']

# --- AGRUPANDO O LUCRO TOTAL POR PRODUTO ---
lucro_por_produto = df.groupby('Produto')['Lucro_Liquido'].sum().sort_values(ascending=False)

# --- CRIANDO O GRÁFICO ---
plt.figure(figsize=(10, 6)) # Define o tamanho da imagem

# Plotando as barras
lucro_por_produto.plot(kind='bar', color='#4CAF50', edgecolor='black')

# Adicionando Títulos e Rótulos (O "Storytelling" dos dados)
plt.title('Lucro Total por Categoria de Produto', fontsize=16, fontweight='bold')
plt.xlabel('Produto', fontsize=12)
plt.ylabel('Lucro Total (R$)', fontsize=12)
plt.xticks(rotation=45) # Rotaciona os nomes para não ficarem em cima uns dos outros
plt.grid(axis='y', linestyle='--', alpha=0.7) # Adiciona linhas de grade suaves

# Exibindo o gráfico diretamente no Notebook
plt.show()
```

Análise de Resultados (Business Intelligence)

->Produto Estrela: A categoria Reborn Luxo lidera o lucro total, indicando alta demanda ou margens muito saudáveis.

->Eficiência de Custos: Mesmo com custos de produção elevados, o valor agregado compensa a operação.
